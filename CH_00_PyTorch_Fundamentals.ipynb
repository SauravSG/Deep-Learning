##### Matrix multiplication
##### - Two main ways of performing multiplication in neural networks and deep learning:
1. Element-wise multiplication
2. Matrix multiplication

There are 2 main rules that performing matrix multiplication needs to satisfy:
1. The **inner dimensions** must match:

* `(3, 2) @ (3, 2)` - won't work

* `(2, 3) @ (3, 2)` - work

* `(3, 2) @ (2, 3)` - work.

2. The resulting matrix  has the shape of the **outer dimensions**:
* `(2, 3) @ (3, 2)` -> `(2, 2)`
* `(3, 2) @ (2, 3)` -> `(3, 3)`

````python
# 1. element wise
print(tensor, '*', tensor)
print(f'Equals: {tensor * tensor}')
````

````python
# 2. Matrix multiplication although we are doing vector multiplication
torch.matmul(tensor, tensor)
````

###### One of the most common errors in DL: shape errors

````python
# will throw inner since the inner dimentions mismatch
torch.matmul(torch.rand(3, 2), torch.rand(3, 2))
````

````python
# will work
torch.matmul(torch.rand(3, 2), torch.rand(2, 3))
````

````python
# shapes for matrix multiplication
tensor_A = torch.tensor([[1, 2],
                        [3, 4],
                        [5, 6]])
tensor_B = torch.tensor([[7, 10],
                        [8, 11],
                        [9, 12]])

torch.mm(tensor_A, tensor_B) #.mm is same as .matmul. More like an alias.
````

To fix our tensor shape issues, we can manipulate one of our matrix using **Transpose**.

A **Transpose** switches the axes or dimensions of a given tensor.


````python
print(f"Original tensor: \n {tensor_B} \n and it's shape is {tensor_B.shape}\n")
print(f'After doing Transpose: \n {tensor_B.T} \n')
print(f'Now the shape after transpose is: \n {tensor_B.T.shape}')
````

````python
print(f'After the Transpose, the Matrix Multiplication will now throw error due to same inner dimensions \n')
print(torch.mm(tensor_A, tensor_B.T), '\n')
print(f'Now the outer dimension will match as well: {torch.mm(tensor_A, tensor_B.T).shape}')
````

###### Reshaping, stacking, squeezing, unsqueezing

* Reshaping - reshapes an i/p tensor at defined shape
* View - return a view of an i/p tensor of certain shape but keep the same memory as the original tensor
* stacking - combine multiple tensors on top (vstack) or side-to-side (hstack)
* squeeze - removes all `1` dimensions from tensor
* Unsqueeze - add a `1` dimension to a target tensor
* Permute - Return a view of the i/p with dimensions permuted (swapped) in a certain way

````python
new_x = torch.arange(1, 10)
print(new_x, new_x.shape)
````

````python
# 1. reshaping
# add an extra dimension
new_x_reshaped = new_x.reshape(1, 7)
print(new_x_reshaped) # will throw error: shape '[1, 7]' is invalid for input of size 9
````

````python
new_x_reshaped = new_x.reshape(9, 1)
print(new_x_reshaped, new_x_reshaped.shape)
````

**Note:** The reshaping works as long as the new shape is multiple of the original one

````python
# change the view
y = new_x.view(1, 9)
print(y, y.shape)
````

````python
# changing y, changes x because a view of a tensor shares the same memory as the original one
y[:, 0] = 5
print(new_x, '\n' ,y)
````

````python
# stack tensors on top of each other based on the dimensions 0 or 1 or even 2 if the original tensor is compatible
x_stacked = torch.stack([new_x, new_x, new_x, new_x], dim = 0)
print(x_stacked, x_stacked.shape)
````

````python
# .vstack() - this stacks the tensors on top of other
y_vstack = torch.vstack([new_x, new_x])
print(y_vstack, y_vstack.shape)
````

````python
# .hstack() - this stacks the tensors side to side
x_hstack = torch.hstack([new_x, new_x])
print(x_hstack, x_hstack.shape)
````

````python
new = torch.tensor([[1, 2, 3],
                    [4, 5, 6]])
print(f'Here, in case of vstack, the tensors will be on top of another: \n{torch.vstack([new, new])}.\n And the shape would be {torch.vstack([new, new]).shape} \n')
print(f'Here, in case of hstack, the tensors will be side to side: \n{torch.hstack([new, new])}.\n And the shape would be {torch.hstack([new, new]).shape}')
````

````python
# squeezing - torch.squeeze() removes all single dimensions from a target tensor
x = torch.arange(1, 11)
print(x, x.shape, '\n')
x_reshape = x.reshape(2, 5)
print(x_reshape, x_reshape.shape)
````

````python
print(x.squeeze())
````

````python
xx = torch.tensor([[1, 2, 3, 4, 5]])
print(xx, xx.shape, '\n')
print(xx.squeeze(), xx.squeeze().shape)
````

Correct condition for squeeze():

**squeeze()** removes any dimension whose size is 1, regardless of position.
Can be squeezed:
`(1, 10)   → (10,)`
`(10, 1)   → (10,)`
`(1, 10, 1) → (10,)`
`(3, 1, 5)   → (3, 5)`

Cannot be squeezed:
`(2, 5)`

It is not limited to just:

(1, any_col) or
(any_row, 1)

````python
#unsqueeze() - adds a single dim to a target tensor at a specific dim

print(f'Previous target: {xx} \n with shape: {xx.shape} \n')

xx_unsqueezed = xx.unsqueeze(dim = 0)
print(f'New tensor: {xx_unsqueezed} \n with shape: {xx_unsqueezed.shape}')
````

````python
#permute - rearranges the dimensions of a target tensor in a specific order

x_original = torch.randn(size=(244, 244, 3)) # [height, width, colur_channels]
print(f'shape of original tensor: \n {x_original.shape} \n')
# permute the original tensor to rearrange the axis (or dim) order

x_permuted = x_original.permute(2, 0, 1)
print(f'New order: \n {x_permuted.shape}')
````

##### Reproducibility: Trying to take random out of random

Getting the same results every time you run the same code.

Without reproducibility:

- Training the same model twice → different weights

- Different loss values

- Hard to debug

- Hard to compare experiments

With reproducibility:

- You can trust comparisons

- You can debug reliably

- Results are repeatable

In short, how a NN learns:
`start with random number -> tensor operations -> update random numbers to try and make them better representations of data -> again -> again...`

To reduce randomness in NN and PyTorch comes the concept of a **random seed**.

Essentially, what the random seed does is the flavour the randomness.

````python
torch.rand(3, 3)
````

````python
# create two random tensors
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(random_tensor_A, '\n')
print(random_tensor_B, '\n')
print(random_tensor_A == random_tensor_B)
````

````python
# making random but reproducible tensors

# set the random seed
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)

random_tensor_C = torch.rand(3, 4)
random_tensor_D = torch.rand(3, 4)
print(random_tensor_C, '\n')
print(random_tensor_D, '\n')
print(random_tensor_C == random_tensor_D)
````

````python
RANDOM_SEED = 42  # it's like a inner state and the tensors will start from the same state

torch.manual_seed(RANDOM_SEED)

random_tensor_C = torch.rand(3, 4)

torch.manual_seed(RANDOM_SEED) # generator is reset back to the same starting point. Previous random state is discarded.

random_tensor_D = torch.rand(3, 4)
print(random_tensor_C, '\n')
print(random_tensor_D, '\n')
print(random_tensor_C == random_tensor_D)
````

torch.manual_seed(42)
- Generator is reset back to the same starting point.
- Previous random state is discarded
